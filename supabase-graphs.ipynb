## Example 2: Query Industrial Market Data

Let's query industrial vacancy data and create a chart.

````python
# Query industrial market data
# Replace 'market_tables_industrial' with your actual table name

response = supabase.table('market_tables_industrial') \
    .select('*') \
    .gte('quarter', '2020-01-01') \
    .order('quarter', desc=False) \
    .execute()

# Convert to pandas DataFrame
df = pd.DataFrame(response.data)

print(f"✓ Loaded {len(df)} rows")
print(f"\nColumns: {df.columns.tolist()}")
print(f"\nFirst few rows:")
df.head()
````

````python
# Generate Aquila-styled line chart
# Adjust column names based on your actual data

# Use 'quarter' as the x-axis directly (as "YYYY Q#" format, e.g., "2020 Q1")
# Do not convert to datetime/PeriodIndex since format is "2020 Q1" and not parseable by pandas
if 'quarter' in df.columns:
    df['quarter'] = df['quarter'].astype(str).str.strip()

# Create the chart
fig = aquila_styled_line_chart(
    df[df['property_type'] == 'Industrial'],
    x='quarter',
    y='total_vacancy_rate',  # Adjust to your column name
    color='submarket_name',   # Adjust to your column name
    title='Industrial Vacancy Rate by Submarket (Supabase Data)',
    height=800
)

# Format y-axis as percentage
fig.update_yaxes(tickformat='.1%')

# Display the chart
fig.show()

# Export to HTML
fig.write_html('charts/supabase_industrial_vacancy.html')
print("\n✓ Chart saved to charts/supabase_industrial_vacancy.html")
````

````python
# Example: Query with multiple filters
response = supabase.table('your_table_name') \
    .select('column1, column2, column3') \
    .eq('property_type', 'Industrial') \
    .gte('quarter', '2023-01-01') \
    .lte('quarter', '2025-12-31') \
    .order('quarter', desc=False) \
    .execute()

df_filtered = pd.DataFrame(response.data)
print(f"Filtered data: {len(df_filtered)} rows")
````

## Supabase Query Features

Here are common Supabase query methods:

```python
# Select specific columns
.select('col1, col2, col3')

# Filters
.eq('column', 'value')         # Equal to
.neq('column', 'value')        # Not equal to
.gt('column', value)           # Greater than
.gte('column', value)          # Greater than or equal
.lt('column', value)           # Less than
.lte('column', value)          # Less than or equal
.like('column', '%pattern%')   # Pattern matching
.ilike('column', '%pattern%')  # Case-insensitive pattern
.is_('column', None)           # IS NULL
.in_('column', [val1, val2])   # IN list

# Ordering
.order('column', desc=True)    # Sort descending
.order('column', desc=False)   # Sort ascending

# Limiting
.limit(100)                    # Limit rows returned
.range(0, 99)                  # Range of rows

# Execute
.execute()                     # Run the query
```

## Troubleshooting

If you get connection errors:
1. Verify `SUPABASE_URL` and `SUPABASE_KEY` in `aquila_graph.env`
2. Check your Supabase dashboard for correct credentials
3. Ensure your table names match exactly (case-sensitive)
4. Verify your API key has the necessary permissions

For RLS (Row Level Security) issues:
- Make sure your API key has access to the tables
- Check RLS policies in Supabase dashboard
- Consider using service_role key for unrestricted access (be careful!)
